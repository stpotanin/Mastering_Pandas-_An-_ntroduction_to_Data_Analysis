Создайте `pd.Series`, содержащий значения `'10', '20', '30'` c типом  `float`. Выведите его на экран. Затем создайте новый `pd.Series`, который будет содержать элементы первого `Series`, увеличенные вдвое. Выведите его на экран.

```python
import pandas as pd

a = pd.Series([10, 20, 30], dtype='float')
print(a)

b = a * 2
print(b)
```
