Создайте pd.Series, содержащий значения `1.0, 2.0, 3.0`. Примените к каждому элементу функцию $\sqrt{exp( x ) - ln (x) exp(x)-ln(x)}$​ и сохраните результат, округленный до двух знаков после запятой в новый `pd.Series`. Распечатайте обе `Series`.

```python
import pandas as pd
import numpy as np

a = pd.Series([1.0, 2.0, 3.0])
print(a)

f = lambda x: (np.exp(x) - np.log(x))**.5

b = f(a)
print(round(b, 2))
```
