# Создать объект Series из списка дат `['2022-01-01', '2022-01-02', '2022-01-03']`, вывести его на печать.

```python
import pandas as pd

print(pd.Series(['2022-01-01', '2022-01-02', '2022-01-03']))
```

```python
import pandas as pd

print(pd.Series(['2022-01-01', '2022-01-02', '2022-01-03'],
                dtype='datetime64[ns]'))
```
