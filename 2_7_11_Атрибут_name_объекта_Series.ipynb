На вход поступают две строки. В первой находятся даты в формате `'YYYY-MM-DD'`, разделенные пробелом, во второй - значению температуры воздуха в указанные даты.

Создайте на основе этих строк объект Series, в котором в качестве меток выступают даты (тип данных - `datetime64`), а значениями являются температуры (тип данных `int16`). Назовите объект `'температура'`, а его индексы `'дата'`.

Выведите исходный объект и его выборку, отвечающую условию - температура воздуха выше 5 градусов, на печать.

```python
import pandas as pd

s = pd.Series(index=pd.Index([pd.to_datetime(x) for x in input().split()],
                             name='дата'),
              data=input().split(),
              dtype='int16',
              name='температура')
print(s)

selection = s[s > 5]
print(selection)

# 2022-01-01 2022-04-04 2022-05-05 2022-07-23 2022-09-28
# -9 4 8 19 22
```
