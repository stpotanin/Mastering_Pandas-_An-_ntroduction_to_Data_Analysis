На вход поступает строка, содержащая информацию о продажах различных товаров за месяц. Каждый товар представлен в формате "наименование-количество", разделенные пробелом. Необходимо создать объект Series, где в качестве меток будут наименования товаров, а значениями - количество проданных единиц. Назовите объект "продажи", а индексы "товар".

Выведите на печать объект и его выборку - в которой содержатся продажи с нечетным значением. Назовите выборку - "нечетные продажи".

```python
import pandas as pd

index, data = zip(*((x.split('-')) for x in input().split()))

df = pd.Series(data=[int(x) for x in data],
               index=index,
              name='продажи')
df.index.name = 'товар'

print(df)

selection = df[df.values % 2 == 1]
selection.name = 'нечетные продажи'

print(selection)

# булавки-25 нитки-12 иголки-34 пяльца-23
```
