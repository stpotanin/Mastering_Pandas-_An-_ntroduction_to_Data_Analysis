Существуют два списка `books` и `prices`  с названиями книг и их ценами.

Необходимо создать на их основе объект Series с названиями книг в качестве меток и ценами в качестве значений.  

Выведите получившийся объект `Series` на печать.

Далее на вход поступают две строки. В первой - название книги, во второй - новая цена на нее.

Замените цену этой книги в объекте `Series` и выведите его на печать.

```python
import pandas as pd

books = ['Преступление и наказание', 'Мастер и Маргарита', '451 градус по Фаренгейту', '1984', 'Убить пересмешника', 'Гарри Поттер и философский камень']
prices = [499, 599, 399, 299, 349, 699]

df = pd.Series(prices, books)
print(df)

new_book, new_price = input(), int(input())

df[new_book] = new_price
print(df)

# 451 градус по Фаренгейту
# 350
```
