Создайте объект Series с элементами [10, 20, 30, 40] и типом данных 'float64', назначьте индексы ['a', 'b', 'c', 'd']. Выведите результат на печать. Умножьте все элементы на 2, выведите обновленный объект на печать. Вычислите среднее значение элементов и выведите его на печать.

```python
import pandas as pd

df = pd.Series([10, 20, 30, 40],
               index=['a', 'b', 'c', 'd'],
               dtype='float64')

print(df)

df = df * 2

print(df)
print(df.mean())
```

```python
import pandas as pd
import math

df = pd.Series([1, 2, 3])

df_exp = math.exp(df)
print(df_exp)
```
