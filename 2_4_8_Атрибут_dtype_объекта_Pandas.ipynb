Создайте объект `Series` с элементами [10, 20, 30, 40, 50] и задайте ему индексы ['a', 'b', 'c', 'd', 'e'] и тип `int32`. Распечатайте объект `Series`. Затем извлеките и распечатайте значение элемента с индексом `'c'`.

```python
import pandas as pd

df = pd.Series([10, 20, 30, 40, 50],
                index=['a', 'b', 'c', 'd', 'e'],
                dtype='int32')

print(df)
print(df['c'])
```
