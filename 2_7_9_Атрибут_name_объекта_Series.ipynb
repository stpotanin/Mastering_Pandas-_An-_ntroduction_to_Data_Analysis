Дан список books, в котором содержатся названия книг и список `prices` с ценами этих книг.

Создайте объект `Series`, в котором в качестве меток выступает список `books`, а в качестве значений их цена, тип данных  - `uint16`. Назовите объект `'Цены на книги'`, а объект index этой `Series` - `'Название'`.

Итоговый `Series` выведите на печать.

```python
import pandas as pd

books = ['Оно', 'Она', 'Они', 'Он']
prices = [120, 240, 34, 333]

s = pd.Series(prices, index=books, dtype='uint16', name='Цены на книги').rename_axis('Название')
print(s)
```

```python
# @title вариант

import pandas as pd

books = ['Оно', 'Она', 'Они', 'Он']
prices = [120, 240, 34, 333]

s = pd.Series(prices, index=pd.Index(books, name='Название'), dtype='uint16', name='Цены на книги')
print(s)
```
