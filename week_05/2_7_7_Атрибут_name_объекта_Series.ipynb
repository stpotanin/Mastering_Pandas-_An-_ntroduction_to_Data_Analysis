Создайте объект Series, содержащий значения `["a", "b", "c", "d"]` типа `'string'`, и задайте ему имя `"my_series"`. Выведите результат на печать.

```python
import pandas as pd

s = pd.Series(["a", "b", "c", "d"], name="my_series", dtype='string')
print(s)
```
