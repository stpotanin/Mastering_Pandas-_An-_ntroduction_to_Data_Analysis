На вход поступают две строки. В первой находятся даты в формате `'YYYY-MM-DD'`, разделенные пробелом, во второй - значению температуры воздуха в указанные даты.

Создайте на основе этих строк объект `Series`, в котором в качестве меток выступают даты (тип данных - `datetime64`), а значениями являются температуры (тип данных `int16`). Назовите объект `'температура'`, а его индексы `'дата'`. Выведите `Series` и его объект `index` на печать.

```python
import pandas as pd

s = pd.Series(index=pd.Index([pd.to_datetime(x) for x in input().split()],
                             name='дата'),
              data=input().split(),
              dtype='int16',
              name='температура')
print(s)
print(s.index)

# 2022-12-03 2023-03-05 2023-06-01
# -12 12 22
```
