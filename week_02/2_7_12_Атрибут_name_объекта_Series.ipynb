На вход поступает строка с названиями фруктов и овощей, разделенная пробелами.

Создайте на ее основе объект `Series` (тип данных - `string`), назовите его 'ягоды',  в качестве меток используйте маленькие буквы латинского алфавита.

Выведите объект на печать.

Далее создайте и выведите на печать объект - выборку исходного объекта, в которой названия начинаются с буквы `'б'`. Назовите выборку - `'ягоды на букву б'`.

```python
import pandas as pd
import string

df = pd.Series(input().split(),
              name='ягоды',
               dtype='string')
df.index = list(string.ascii_lowercase[:len(df)])

print(df)

selection = df[[x[0] == 'б' for x in df]]
# selection = df[df.str.startswith('б')]
selection.name = 'ягоды на букву б'

print(selection)

# ананас апельсин баклажан банан клубника бурак картофель
```

```python
# @title Бесконечный генератор колонок в стиле Excel
import pandas as pd
import string

def excel_style_col_names():
    """Бесконечный генератор колонок в стиле Excel: A, B, ..., Z, AA, AB, ..."""
    n = 1
    while True:
        for c in _index_to_letters(n):
            yield c
        n += 1

def _index_to_letters(n):
    """Генерирует все комбинации длины n"""
    if n == 1:
        yield from string.ascii_lowercase
    else:
        for prefix in _index_to_letters(n - 1):
            for suffix in string.ascii_lowercase:
                yield prefix + suffix

df = pd.Series(input().split(),
              name='ягоды',
               dtype='string')
gen = excel_style_col_names()
df.index = [next(gen) for _ in range(len(df))]

print(df)

selection = df[df.str.startswith('б')] #
selection.name = 'ягоды на букву б'

print(selection)
```

```python
# @title Библиотека openpyxl уже умеет генерировать индексы в стиле Excel
import pandas as pd
from openpyxl.utils import get_column_letter

df = pd.Series(input().split(),
              name='ягоды',
               dtype='string')

# get_column_letter принимает номер колонки (начиная с 1)
df.index = [get_column_letter(i + 1) for i in range(len(df))]

print(df)

selection = df[[x[0] == 'б' for x in df]]
selection.name = 'ягоды на букву б'

print(selection)
```
