Есть два списка, содержащие информацию о продажах различных товаров в течение месяца. В первом списке `products` - наименование, во втором `sales` - количество продаж. Необходимо создать объект `Series`, где в качестве меток будут наименования товаров, а значениями - количество проданных единиц. Назовите объект `"продажи за месяц"`, а индексы `"товар"`. Выведите объект на экран.

Далее выведите на экран наименее продаваемый товар за этот период, только название. Используйте атрибут `index`.

```python
import pandas as pd

products = ['iPhone 13', 'Samsung Galaxy S21', 'Google Pixel 6', 'OnePlus 9 Pro', 'Xiaomi Mi 11', 'Oppo Find X3 Pro', 'Sony Xperia 1 III', 'LG Wing']
sales = [50000, 35000, 15000, 12000, 9000, 7500, 5000, 2000]

df = pd.Series(data=sales,
               index=products,
               name='продажи за месяц')
df.index.name = 'товар'

print(df)
print(*df[df == min(df)].index)
```
