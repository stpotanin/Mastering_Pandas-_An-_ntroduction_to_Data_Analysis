Существуют два массива: `dates` (содержит даты в строковом представлении в формате `'YYYY-MM-DD'`) и `temperatures` (содержит температуры воздуха в соответствующие даты).  Создайте на их основе объект `Series` с датами в качестве меток и температурой в качестве значений.

Далее на вход поступает целое число `n` - номер месяца, вам необходимо сделать выборку из созданного объекта `Series` и вывести температуры в указанный месяц.

```python
import pandas as pd
import numpy as np
from datetime import date, timedelta

np.random.seed(33)
dates = list(map(str,[date(2019, 1, 4) + timedelta(days=i) for i in range(360)]))
temperatures = np.random.randint(-10, 35, len(dates))

df = pd.Series(temperatures, dates)

n = int(input())

month_n = df[pd.to_datetime(df.index).month == n]
print(month_n)
```
