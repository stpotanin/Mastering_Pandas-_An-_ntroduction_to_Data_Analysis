# На вход поступает строка слов, разделенных пробелом. Создайте на ее основе Series s с текстовыми значениями типа string.

# Далее во второй строке поступает буква.

# Получите новый Series, содержащий только значения исходной Series, которые начинаются с заданной буквы.

# Обе Series выведите на печать.

```python
import pandas as pd

a = pd.Series(input().split(), dtype='string')
print(a)

letter = input()
b = a[[x[0] == letter for x in a]]
print(b)
```
