# На вход поступает две строки. В первой строке через пробел следуют даты, во второй строке через пробел следуют цены на акции некоторой компании в соответствующие даты. Создайте объект Series, в котором индексами являются даты, а значениями  - цены.

# Далее в третьей строке через пробел поступают даты, вам необходимо вывести на печать цены акций в указанные даты в строку через пробел.

```python
import pandas as pd

date, price = input().split(), input().split()
df = pd.Series([int(x) for x in price], date)

print(*df[input().split()].values)
```
