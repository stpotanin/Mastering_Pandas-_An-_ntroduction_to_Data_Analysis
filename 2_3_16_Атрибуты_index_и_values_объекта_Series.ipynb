Существуют два списка `dates` с датами и  `temperatures` с температурами по Фаренгейту. Необходимо создать на их основе объект `Series` с датами в качестве меток и температурами в качестве значений. Выведите его на печать. Далее необходимо пересчитать значения температуры в созданном объекте и перевести их в градусы Цельсия по формуле

(температура в Фаренгейтах - 32) * 5/9.

Выведите обновленное значение Series на печать через пустую строку.

```python
import pandas as pd

dates = ['2022-06-04', '2022-06-05', '2022-06-06', '2022-06-07', '2022-06-08', '2022-06-09', '2022-06-10']
temperatures = [68, 72, 75, 70, 68, 73, 79]

df = pd.Series(temperatures, index=dates)
print(df, '\n')

df = (df - 32) * 5 / 9
print(df)
```
