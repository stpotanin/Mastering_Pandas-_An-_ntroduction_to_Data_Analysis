# На вход поступает строка дат в формате 'YYYY-MM-DD', разделенных пробелом. Создайте на ее основе Series с датами типа datetime64.

# Получить новый Series, содержащий только даты, которые попадают вo временной период '2021-03-18' - '2021-09-22' (невключительно).

# Обе Series выведите на печать.

```python
import pandas as pd

a = pd.Series(input().split(), dtype='datetime64[ns]')
print(a)

b = a['2021-03-18' < a][a < '2021-09-22']
print(b)

# 2021-01-01 2021-02-21 2021-03-18 2021-05-02 2021-08-11 2021-12-12
```

```python
# @title вариант
import pandas as pd

a = pd.Series(input().split(), dtype='datetime64[ns]')
print(a)

b = a[(a > '2021-03-18') & (a < '2021-09-22')]
print(b)

```
