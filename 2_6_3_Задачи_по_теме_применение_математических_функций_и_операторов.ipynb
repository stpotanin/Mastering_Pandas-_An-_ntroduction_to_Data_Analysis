Создайте `pd.Series` из списка `[1, 2, 3, 4, 5]` и назначьте в качестве меток первые маленькие латинские буквы алфавита. Выведите `Series` на экран. Используя условный оператор, выведите на экран далее только те значения, которые меньше `3`.

```python
import pandas as pd

a = pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e'])
i = a < 3
print(a)
print(a[i])
```
