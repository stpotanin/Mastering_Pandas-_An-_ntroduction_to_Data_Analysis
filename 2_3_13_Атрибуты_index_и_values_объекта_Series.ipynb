# На вход в первой строке поступают имена студентов через пробел, во второй строке - их оценки.

# Создать на их основе объект Series с именами в качестве меток и оценками в качестве значений.

# Далее в третьей строке поступают имена через пробел. Необходимо вывести оценки указанных студентов в строку через пробел.

```python
import pandas as pd

names, scores = input().split(), input().split()
df = pd.Series(scores, names)

print(*df[input().split()].values)

# Иван Олег Татьяна Марина Алексей Андрей
# 65 61 83 68 69 99
# Олег Татьяна          61 83
```
