Существует словарь `data`, в котором ключи - строки, а значения списки. Создайте на его основе DataFrame, так чтобы ключи были названиями колонок, а значения выбирались из соответствующих списков. Распечатайте DataFrame.

```python
import pandas as pd

data = {
    'name': ['Alice', 'Bob', 'Charlie', 'David'],
    'age': [25, 32, 19, 47],
    'gender': ['female', 'male', 'male', 'male'],
    'city': ['New York', 'Los Angeles', 'Chicago', 'Houston']
}

df = pd.DataFrame(data)

print(df)
```
