Дан словарь `countries`, в котором ключами являются названия стран, а значениями численность их населения.

Создать на его основе объект `Series`, в котором названия стран выступают в качестве меток, а население - значений. Дайте объекту название - `'численность населения'`, а индексу название - `'страна'`.

Распечатайте получившуюся `Series`.

```python
import pandas as pd

countries = {'Индия':1416, 'Китай': 1411, 'США': 336, 'Индонезия': 287}

s = pd.Series(countries, name="численность населения")
s.index.name = 'страна'
# s = pd.Series(countries, name="численность населения").rename_axis('страна')

print(s)
```
