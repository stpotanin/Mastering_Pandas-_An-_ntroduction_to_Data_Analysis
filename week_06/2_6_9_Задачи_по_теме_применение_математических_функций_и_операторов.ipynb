На вход поступает строка целых чисел, разделенных пробелом. Создайте на ее основе `Series` с беззнаковым целым типом `uint32`. Получите новый `Series`, содержащий только нечетные значения из исходного.  Обе `Series` выведите на печать.

```python
import pandas as pd

a = pd.Series([int(x) for x in input().split()], dtype='uint32')
print(a)

b = a[a % 2 == 1]
print(b)
```
