# На вход поступает строка слов, разделенных пробелом. Создайте на ее основе Series s с текстовыми значениями типа string.

# Получите новый Series, содержащий только значения исходной Series, которые начинаются на букву 'а' и заканчиваются на букву 'к'.

# Обе Series выведите на печать.

```python
import pandas as pd

a = pd.Series(input().split(), dtype='string')
print(a)

b = a.loc[[x[0] == 'а' and x[-1] == 'к' for x in a]]
print(b)

# абрек аул барнаул кит арнлк
```
