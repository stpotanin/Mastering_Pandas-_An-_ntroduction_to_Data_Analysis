На вход поступает строка целых чисел, разделенных пробелом. Создайте на ее основе `Series` s с числовыми значениями типа `int8`.  

Выведете `Series` на печать. Получите новый `Series`, содержащий только значения, которые делятся на три без остатка. Выведите на печать.

```python
import pandas as pd

s = pd.Series([int(x) for x in input().split()], dtype='int8')
print(s)

s_new = s[s % 3 == 0]
print(s_new)
```
