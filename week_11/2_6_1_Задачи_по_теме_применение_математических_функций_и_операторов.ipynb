Создайте `pd.Series` из списка `[1, 2, 3, 4, 5]` и `pd.Series` из списка `[2, 4, 6, 8, 10]` с типом `int16`. Найдите сумму и произведение этих серий. Выведите результаты на экран, сначала сумму, потом произведение.

```python
import pandas as pd

a = pd.Series([1, 2, 3, 4, 5], dtype='int16')
b = pd.Series([2, 4, 6, 8, 10], dtype='int16')

print(a + b)
print(a * b)
```
