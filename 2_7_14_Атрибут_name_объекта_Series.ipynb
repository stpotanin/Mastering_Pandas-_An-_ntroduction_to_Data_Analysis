На вход поступает строка, содержащая информацию о количестве посетителей сайта за каждый день месяца. Каждый день представлен в формате "день:количество", разделенные пробелом. День представлен в формате "YYYY-MM-DD".

Необходимо создать объект Series, где в качестве меток будут даты (формат datetime64), а значениями - количество посетителей (формат uint32). Назовите объект "посещаемость", метки - "дата".

Выведите объект Series и его index на печать.

```python
import pandas as pd

index, data = zip(*((x.split(':')) for x in input().split()))

df = pd.Series(data=[int(x) for x in data],
               index=[pd.to_datetime(x) for x in index],
              name='посещаемость',
               dtype='uint32')
df.index.name = 'дата'

print(df)
print(df.index)

# 2023-01-05:234 2023-02-23:345 2023-03-12:239
```
