Создайте `pd.Series` из списка `[100, 200, 300, 400, 500]` и `pd.Series` из списка `[50, 100, 150, 200, 250]`. Найдите разность и частное этих серий. Результаты выведите на экран, сначала разность, потом частное.

```python
import pandas as pd

a = pd.Series([100, 200, 300, 400, 500])
b = pd.Series([50, 100, 150, 200, 250])

print(a - b)
print(a / b)
```
