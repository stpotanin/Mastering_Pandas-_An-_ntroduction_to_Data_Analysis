На вход поступает строка целых чисел, разделенных пробелом. Создайте на ее основе `Series` s с числовыми значениями типа `int32`. Выведите ее на печать.

Во второй строке далее подаются два числа `a < b`, разделенные пробелом. Получите новый `Series`, содержащий только значения, которые находятся в интервале между `a` и `b` (оба невключительно). Выведите его на печать.

```python
import pandas as pd

s = pd.Series([int(x) for x in input().split()], dtype='int32')
print(s)

a, b = [int(x) for x in input().split()]

s_new = s[a < s][s < b]
print(s_new)
```
