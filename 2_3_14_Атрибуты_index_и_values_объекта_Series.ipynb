# На вход поступаю две строки. В первой названия стран через пробел, во второй население в этих странах.

# Необходимо создать на их основе объект Series с названиями стран в качестве меток и населением в них в качестве значений. Тип данных для населения должен быть `float64`.

# Далее в третьей строке поступает строка, в которой сначала идет название страны и далее новое значение ее населения. Необходимо вывести обновленное значение созданного объекта Series.

```python
import pandas as pd

countries, populations = input().split(), [float(x) for x in input().split()]
df = pd.Series(populations, countries)

new_countriy, new_population = input().split()

df[new_countriy] = float(new_population)

print(df)

# Россия США Китай Индия Япония Германия Великобритания Франция Италия Канада Южная_Корея Австралия
# 144.5 328.2 1393 1366 126.5 83.1 66.7 66.99 60.4 37.59 51.78 25.4
# Германия 82
```
