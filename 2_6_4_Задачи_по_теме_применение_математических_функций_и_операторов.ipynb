Создайте pd.Series, содержащий значения 10, 20, 30. Измените тип элементов в этом Series на тип bool таким образом, чтобы все числа больше 15 были преобразованы в True, а все остальные - в False. Выведите исходный и итоговый Series на экран.

```python
import pandas as pd

a = pd.Series([10, 20, 30])
print(a)

a = a > 15
print(a)
```
