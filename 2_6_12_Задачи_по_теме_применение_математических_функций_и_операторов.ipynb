На вход поступает строка целых чисел, разделенных пробелом. Создайте на ее основе `Series` s с числовыми значениями  типа `float32`.

Получите новый `Series`, содержащий только значения, которые больше среднего значения в `s`.

Выведите обе Series на печать.

```python
import pandas as pd
from numpy import mean

s = pd.Series([int(x) for x in input().split()], dtype='float32')
print(s)

s_new = s[s > s.mean()]
print(s_new)
```
