Существует список списков `data` с данными и список `cols` с колонками, создайте на его основе `DataFrame` и выведите его на печать.

```python
import pandas as pd

data = [['John', 28, 'New York'],
        ['Kate', 23, 'Chicago'],
        ['Mike', 32, 'San Francisco']]
cols = ['Name', 'Age', 'City']

df = pd.DataFrame(data, columns=cols)

print(df)
```
