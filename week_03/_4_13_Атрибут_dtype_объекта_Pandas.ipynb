На вход поступает строка чисел, разделенных пробелом, содержащая данные о количестве продаж товаров в магазине за определенный период времени. Создайте объект Series на ее основе, используйте тип данных `'int32'`.

Выведите объект `Series` на печать.

Далее во второй строке поступает последовательность целых чисел - меток, выведите выборку из объекта `Series` по этим меткам.

```python
import pandas as pd

df = pd.Series([int(x) for x in input().split()], dtype='int32')

print(df)
print(df[[int(x) for x in input().split()]])

# 16 41 42 19 17 19 29 19 26 24 30 33 34 16 38
# 2 3 7 4 0 1
```
