Создайте объект Series с элементами `['red', 'green', 'blue']` и типом данных `'string'`. Назначьте индексы. Выведите его на печать. Используя индексацию, извлеките элемент с индексом 'b' и выведите его на печать.

```python
import pandas as pd

df = pd.Series(['red', 'green', 'blue'],
               index=['a', 'b', 'c'],
               dtype='string')

print(df)
print(df['b'])
```
